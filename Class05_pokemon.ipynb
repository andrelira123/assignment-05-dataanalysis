```python
print(pokemon_df.hp.min())
print(pokemon_df.hp.max())
```

```python
strongest_pokemon = pokemon_df.loc[pokemon_df.hp == 255,'name']
weakest_pokemon = pokemon_df.loc[pokemon_df.hp == 1,'name']

print(strongest_pokemon)
print(weakest_pokemon)
```

```python
plt.hist(pokemon_df.hp, bins=25)
plt.xlabel("HP")
plt.ylabel("Number of Pokemon")
plt.title("Distribution of HP among Pokemon")
plt.axvline(pokemon_df.hp.mean(), color="red",linestyle="dashed",linewidth=2)
plt.axvline(pokemon_df.hp.median(), color="yellow",linestyle="dashed",linewidth=2)
plt.show()
```

```python
type_groupings = pokemon_df.groupby('type1')

water_df = type_groupings.get_group('water')
fire_df = type_groupings.get_group('fire')
water_df.head()
fire_df.head()
```

```python
plt.hist(water_df.hp,bins=20,label="water")
plt.hist(fire_df.hp, bins=20, label="fire")
plt.legend()
```

```python
correlation = pokemon_df.height_m.corr(pokemon_df.weight_kg)
correlation
```

```python
pokemon_values = pokemon_df[['speed', 'hp', 'sp_attack', 'sp_defense']]
corr_pokemon = pokemon_values.corr()
corr_pokemon
```

```python
plt.matshow(pokemon_values.corr())
plt.xticks(range(pokemon_values.select_dtypes(['number']).shape[1],pokemon_values.select_dty))
plt.yticks(range(pokemon_values.select_dtypes(['number']).shape[1],pokemon_values.select_dty))
cb = plt.colorbar()
cb.ax.tick_params(labelsize=14)
plt.title('Correlation Matrix', fontsize=16);
plt.show()
```
